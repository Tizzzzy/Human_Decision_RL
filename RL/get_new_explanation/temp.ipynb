```python
import os
import json

# Let's check if the file exists in the current directory
files = os.listdir('.')
if 'claude_predictions.json' in files:
    with open('claude_predictions.json', 'r') as f:
        data = json.load(f)
else:
    print("File not found in current directory. Will need to parse from prompt text.")
    data = None

if data:
    correct = sum(1 for item in data if item.get('label') == item.get('model_prediction'))
    total = len(data)
    print(f"Correct: {correct}, Total: {total}, Accuracy: {correct/total:.4f}")
```
